```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
# We'll imagine this is sensor data from various devices.
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0), # Celsius
    'Humidity_Sensor': (30.0, 80.0),   # Percentage
    'Pressure_Sensor': (980.0, 1030.0),# hPa
    'Light_Sensor': (100, 1000)        # Lux
}

# --- Task 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# 1.1 Generate a list of 100 simulated sensor readings.
#    Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value).
#    - timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
#    - device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
#    - device_type: Randomly chosen from DEVICE_TYPES.
#    - location: Randomly chosen from LOCATIONS.
#    - value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
# Store these in a list called 'raw_sensor_data'.

# Your code for Task 1.1 here:
# Task 1.1: Generate 100 simulated sensor readings

raw_sensor_data = []

for i in range(1, 101):
    # Generate timestamp
    timestamp = datetime.datetime.now() - datetime.timedelta(
        minutes=random.randint(0, 10000)
    )
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    # Generate unique device ID
    device_id = f"DEV{i:03d}"

    # Select random device type
    device_type = random.choice(DEVICE_TYPES)

    # Select random location
    location = random.choice(LOCATIONS)

    # Generate value based on device type range
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    # Create tuple and add to list
    reading = (timestamp_str, device_id, device_type, location, value)
    raw_sensor_data.append(reading)

# Display first 5 readings
print("\nFirst 5 Sensor Readings:")
for reading in raw_sensor_data[:5]:
    print(reading)

print(f"\nTotal readings generated: {len(raw_sensor_data)}")


```

```python
# Task 1.2: Create temperature_readings list

temperature_readings = []

for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading

    if device_type == 'Temperature_Sensor':
        temp_dict = {
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }
        temperature_readings.append(temp_dict)

# Display first 5 temperature readings
print("\nFirst 5 Temperature Readings:")
for temp in temperature_readings[:5]:
    print(temp)

print(f"\nTotal Temperature Sensor Readings: {len(temperature_readings)}")
```

```python
# 1.3 Create a set named 'unique_locations' containing all unique locations from 'raw_sensor_data'.

# Your code for Task 1.3 here:
unique_locations = {reading[3] for reading in raw_sensor_data}

print("Unique Locations:", unique_locations)
print("Total Unique Locations:", len(unique_locations))

```

```python
# 1.4 Create a dictionary named 'device_stats' where keys are 'device_type' and values are
#     a tuple: (min_value, max_value) observed for that device type across all readings.
#     Initialize with sensible large/small values.

# Your code for Task 1.4 here:
# Task 1.4: Create device_stats dictionary

# Initialize dictionary with large/small values
device_stats = {}

for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

# Find min and max values for each device type
for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    if value < current_min:
        current_min = value

    if value > current_max:
        current_max = value

    device_stats[device_type] = (current_min, current_max)

# Display results
print("\nDevice Statistics (Min, Max):")
for device_type, stats in device_stats.items():
    print(f"{device_type}: {stats}")
```

```python
# --- Task 2: Function Definitions and Usage ---

# 2.1 Function to get readings by location

def get_readings_by_location(readings_list, target_location):
    """
    Returns all sensor readings for a specified location.
    """
    return [
        reading for reading in readings_list
        if reading[3] == target_location
    ]


# Example Usage
lab1_readings = get_readings_by_location(raw_sensor_data, "Lab1")
print(f"Number of readings from Lab1: {len(lab1_readings)}")


# 2.2 Function to calculate average temperature

def calculate_average_temperature(temp_data_list):
    """
    Calculates and returns the average temperature.
    Returns 0.0 if the list is empty.
    """
    if not temp_data_list:
        return 0.0

    total_temperature = sum(
        reading['temperature_celsius']
        for reading in temp_data_list
    )

    return total_temperature / len(temp_data_list)


# Example Usage
avg_temp = calculate_average_temperature(temperature_readings)
print(f"Average Temperature: {avg_temp:.2f}°C")


# 2.3 Function to generate a formatted report line

def generate_report_line(reading_dict):
    """
    Generates a formatted report string for a temperature reading.
    """
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Example Usage
if temperature_readings:
    report_line = generate_report_line(temperature_readings[0])
    print(report_line)
```

```python
# 4.1 Write temperature readings to file

report_filename = os.path.join(reports_dir, "temperature_report.txt")

try:
    with open(report_filename, "w") as f:
        for reading_dict in temperature_readings:
            f.write(generate_report_line(reading_dict) + "\n")

    print(f"\nTemperature report written successfully to: {report_filename}")

except IOError as e:
    print(f"\nFile writing error: {e}")
```
