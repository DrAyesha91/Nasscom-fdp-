```python
## Norms & Cosine Similarity
w = np.array([3, -4])

print("L1:", np.linalg.norm(w, 1))       # 7.0
print("L2:", np.linalg.norm(w, 2))       # 5.0
print("Linf:", np.linalg.norm(w, np.inf))# 4.0

from sklearn.metrics.pairwise import cosine_similarity

a = np.array([[1, 2, 3]])
b = np.array([[2, 4, 6]])

print(cosine_similarity(a, b))           # [[1.0]]
```

```python
# 3 vectors students
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Student vectors (Maths, Science, English)
u = np.array([85, 90, 80])
v = np.array([80, 88, 78])
w = np.array([60, 65, 70])
print("L1 (A):", np.linalg.norm(u, 1))        # 255.0
print("L2 (A):", np.linalg.norm(u, 2))        # 142.13
print("Linf (A):", np.linalg.norm(u, np.inf)) # 90.0

print("L1 (B):", np.linalg.norm(v, 1))        # 246.0
print("L2 (B):", np.linalg.norm(v, 2))        # 136.10
print("Linf (B):", np.linalg.norm(v, np.inf)) # 88.0

print("L1 (C):", np.linalg.norm(w, 1))        # 195.0
print("L2 (C):", np.linalg.norm(w, 2))        # 109.54
print("Linf (C):", np.linalg.norm(w, np.inf)) # 70.0
a = np.array([u])
b = np.array([v])
c = np.array([w])

print("A vs B:", cosine_similarity(a, b))   # [[0.9996]]
print("A vs C:", cosine_similarity(a, c))   # [[0.9980]]
print("B vs C:", cosine_similarity(b, c))   # [[0.9975]]
```

```python
# 🛠️ Hands-On #2: Matrix Operations in NumPy

## Matrix Creation & Basic Ops
import numpy as np

A = np.array([[2, 1], [0, 3]])
B = np.array([[1, 4], [2, 5]])

print("A + B =\n", A + B)
print("A transpose =\n", A.T)
print("A @ B =\n", A @ B)    # Matrix multiplication
```

```python
## Identity & Inverse
I = np.eye(3)                # 3x3 identity matrix
print("Identity:\n", I)

A = np.array([[2, 1], [1, 3]])
A_inv = np.linalg.inv(A)

print("A inverse:\n", A_inv)
print("A @ A_inv =\n", A @ A_inv)   # Should be identity
```

```python
## Solving Ax = b
# Solve: 2x + y = 5, x + 3y = 7

A = np.array([[2, 1], [1, 3]])
b = np.array([5, 7])

x = np.linalg.solve(A, b)    # More stable than inv(A) @ b

print("Solution:", x)        # [1.6, 1.8]

# Verify: A @ x should equal b
print("Check: A @ x =", A @ x)   # [5.0, 7.0]
```

```python
import numpy as np

# Product prices: Pen, Notebook, Eraser

A = np.array([
    [2, 1, 3],   # Order 1
    [1, 2, 1],   # Order 2
    [3, 1, 2]    # Order 3
])

b = np.array([19, 13, 20])   # Total costs

prices = np.linalg.solve(A, b)

print("Pen, Notebook, Eraser prices:")
print(prices)

# Verification
print("Check:")
print(A @ prices)
```
